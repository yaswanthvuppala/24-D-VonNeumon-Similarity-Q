```python
import os

# 1. Create data directory
os.makedirs("data", exist_ok=True)

# 2. Download from official Hugging Face mirror (~1.28 GB zip)
!wget -c -O data/hpatches-sequences-release.zip "https://huggingface.co/datasets/vbalnt/hpatches/resolve/main/hpatches-sequences-release.zip"

# 3. Extract the zip archive
!unzip -q data/hpatches-sequences-release.zip -d data/

# 4. Remove zip archive to free Kaggle disk space
!rm data/hpatches-sequences-release.zip

# 5. Verify all 116 sequences exist
extracted = os.listdir("data/hpatches-sequences-release")
print(f"Success! Total sequences extracted: {len(extracted)}")
```

```python
import subprocess
import sys

# Evaluate all requested (n, y) configurations. Each run keeps its own results.
window_configs = [(13,7),(15,3)]
for n, y in window_configs:
    print(f"\nRunning HPatches with n={n}, y={y} ({n * n - 1}-D)")
    subprocess.run([
        sys.executable, "scripts/hpatches_eval.py",
        "--max-keypoints", "1024",
        "--n", str(n), "--y", str(y),
        "--norm", "l2", "--matching", "mutual_nn",
        "--output-dir", f"outputs/hpatches_n{n}_y{y}",
    ], check=True)
```

```python
# Auto-downloads ~50 MB from Oxford University and evaluates 8 sequences (40 pairs) per configuration.
import subprocess
import sys

window_configs = [(13,7),(15,3)]
for n, y in window_configs:
    print(f"\nRunning Oxford with n={n}, y={y} ({n * n - 1}-D)")
    subprocess.run([
        sys.executable, "scripts/oxford_eval.py",
        "--max-keypoints", "1024",
        "--n", str(n), "--y", str(y),
        "--norm", "l2",
        "--output-dir", f"outputs/oxford_n{n}_y{y}",
    ], check=True)
```

```python
import os

# 1. Download MegaDepth-1500 test set (~595 MB) if not already extracted
if not os.path.isdir("data/megadepth_test_1500/Undistorted_SfM"):
    print("Downloading MegaDepth-1500 test set (~595 MB)...")
    !wget -c -O data/megadepth_test_1500.tar.gz "https://huggingface.co/lihao20888/traqpoint/resolve/main/megadepth_test_1500.tar.gz"
    !tar -xzf data/megadepth_test_1500.tar.gz -C data/
    !rm -f data/megadepth_test_1500.tar.gz
    print("MegaDepth-1500 test set ready!")
else:
    print("MegaDepth-1500 test set already present!")

# 2. Evaluate configurable QSED descriptor on MegaDepth-1500 (matching RIPE Paper Table 2 Left)
# Evaluates 100 pairs evenly sampled across all scenes & overlap ranges (~45s)
# Note: change --num-pairs 1500 for the complete benchmark run
import subprocess
import sys

window_configs = [(13,7),(15,3)]
for n, y in window_configs:
    print(f"\nRunning MegaDepth with n={n}, y={y} ({n * n - 1}-D)")
    subprocess.run([
        sys.executable, "scripts/megadepth_eval.py",
        "--num-pairs", "100", "--max-keypoints", "1024",
        "--n", str(n), "--y", str(y),
        "--norm", "l2",
        "--output-dir", f"outputs/megadepth_n{n}_y{y}",
    ], check=True)
```

```python
# =============================================================================
# CONSOLIDATED BENCHMARK COMPARISON: RIPE (arXiv:2507.04839) vs PROPOSED QSED-24D
# =============================================================================
import os

reports = [
    ("outputs/hpatches/hpatches_results.txt", "1. HPATCHES BENCHMARK (HOMOGRAPHY AUC@1px, 3px, 5px) - RIPE TABLE 2 RIGHT"),
    ("outputs/oxford/oxford_results.txt", "2. OXFORD AFFINE BENCHMARK (AUC@1px, 3px, 5px ACROSS 8 CHALLENGES)"),
    ("outputs/megadepth/megadepth_results.txt", "3. MEGADEPTH-1500 RELATIVE POSE BENCHMARK (AUC@5°, 10°, 20°) - RIPE TABLE 2 LEFT"),
]

for report_path, title in reports:
    print("=" * 85)
    print(f"   {title}")
    print("=" * 85)
    if os.path.isfile(report_path):
        with open(report_path, "r", encoding="utf-8") as f:
            print(f.read().strip())
    else:
        print(f"  [Info] Results file not found at: {report_path}")
    print("\n")
```
